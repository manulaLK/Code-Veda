```python
import pandas as pd
import numpy as np

df = pd.read_csv('../../Data sets/1) iris.csv')

print("=== RAW DATASET SUMMARY ===")
print("Initial Shape:", df.shape)
print("\nFirst 5 Rows:")
print(df.head())
print("\nMissing Values Count:")
print(df.isnull().sum())
print("\nDuplicate Rows Count:", df.duplicated().sum())
```

```python
df_cleaned = df.drop_duplicates().copy()

numeric_cols = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
for col in numeric_cols:
    if df_cleaned[col].isnull().sum() > 0:
        df_cleaned[col] = df_cleaned[col].fillna(df_cleaned[col].median())

df_cleaned['species'] = df_cleaned['species'].astype(str).str.strip().str.lower()

print("=== CLEANED DATASET SUMMARY ===")
print("Cleaned Shape:", df_cleaned.shape)
print("Duplicates After Cleaning:", df_cleaned.duplicated().sum())
print("Unique Species:", df_cleaned['species'].unique())

df_cleaned.to_csv('cleaned_iris.csv', index=False)
print("\nSaved 'cleaned_iris.csv' successfully!")
```
